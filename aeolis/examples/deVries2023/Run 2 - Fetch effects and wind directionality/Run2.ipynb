```python
#pcolor of transport magnitude with black quiver arrows on top, t=40, 0deg vs 30deg vs 60deg
#cross sections show transport magnitude and the onshore component
#domain is only 10m alongshore, so tile it to cover 100m alongshore for display
t_idx = 35
n_reps = 10  # 10m alongshore tiled 10x -> 100m alongshore display
cross_section_idx = 5  # alongshore index of the cross section to display

y_grid_ext, x_grid_ext = np.mgrid[0:100,0:100]

qs_0_t = np.tile(qs[t_idx,:,:,0], (n_reps, 1))
qn_0_t = np.tile(qn[t_idx,:,:,0], (n_reps, 1))
mag_0 = np.sqrt(qs_0_t**2 + qn_0_t**2)

qs_30_t = np.tile(qs_30[t_idx,:,:,0], (n_reps, 1))
qn_30_t = np.tile(q_30n[t_idx,:,:,0], (n_reps, 1))
mag_30 = np.sqrt(qs_30_t**2 + qn_30_t**2)

qs_60_t = np.tile(qs_60[t_idx,:,:,0], (n_reps, 1))
qn_60_t = np.tile(q_60n[t_idx,:,:,0], (n_reps, 1))
mag_60 = np.sqrt(qs_60_t**2 + qn_60_t**2)

vmax = max(mag_0.max(), mag_30.max(), mag_60.max())

skip = 8  # subsample the quiver grid so fewer, larger arrows are shown

fig = plt.figure(figsize=(9, 6))
gs = fig.add_gridspec(
    2,
    4,
    height_ratios=[1, 0.7],
    width_ratios=[1, 1, 1, 0.08],
    wspace=0.1,
    hspace=0.005,
)
ax = np.empty((2, 3), dtype=object)
ax[0, 0] = fig.add_subplot(gs[0, 0])
ax[0, 1] = fig.add_subplot(gs[0, 1], sharey=ax[0, 0])
ax[0, 2] = fig.add_subplot(gs[0, 2], sharey=ax[0, 0])
ax[1, 0] = fig.add_subplot(gs[1, 0], sharex=ax[0, 0])
ax[1, 1] = fig.add_subplot(gs[1, 1], sharex=ax[0, 1], sharey=ax[1, 0])
ax[1, 2] = fig.add_subplot(gs[1, 2], sharex=ax[0, 2], sharey=ax[1, 0])
colorbar_ax = fig.add_subplot(gs[0, 3])
colorbar_position = colorbar_ax.get_position()
colorbar_ax.set_position([
    colorbar_position.x0,
    colorbar_position.y0 + 0.1 * colorbar_position.height,
    colorbar_position.width,
    0.8 * colorbar_position.height,
])

p0 = ax[0, 0].pcolormesh(x_grid_ext, y_grid_ext, mag_0, cmap='viridis', vmin=0, vmax=vmax, shading='auto')
ax[0, 0].quiver(x_grid_ext[::skip,::skip], y_grid_ext[::skip,::skip],
                qs_0_t[::skip,::skip], qn_0_t[::skip,::skip],
                color='k', scale=vmax*15, width=0.006)
ax[0, 0].set_title('$0^\\circ$')
ax[0, 0].set_xlim((0, 98))
ax[0, 0].set_xlabel('')
ax[0, 0].set_ylabel('alongshore distance [m]')
ax[0, 0].tick_params(axis='x', which='both', labelbottom=False)
ax[0, 0].set_aspect('equal', adjustable='box')

p1 = ax[0, 1].pcolormesh(x_grid_ext, y_grid_ext, mag_30, cmap='viridis', vmin=0, vmax=vmax, shading='auto')
ax[0, 1].quiver(x_grid_ext[::skip,::skip], y_grid_ext[::skip,::skip],
                qs_30_t[::skip,::skip], qn_30_t[::skip,::skip],
                color='k', scale=vmax*15, width=0.006)
ax[0, 1].set_title('$30^\\circ$')
ax[0, 1].set_xlim((0, 98))
ax[0, 1].set_xlabel('')
ax[0, 1].set_ylabel('')
ax[0, 1].tick_params(axis='x', which='both', labelbottom=False)
ax[0, 1].tick_params(axis='y', which='both', labelleft=False)
ax[0, 1].set_aspect('equal', adjustable='box')

p2 = ax[0, 2].pcolormesh(x_grid_ext, y_grid_ext, mag_60, cmap='viridis', vmin=0, vmax=vmax, shading='auto')
ax[0, 2].quiver(x_grid_ext[::skip,::skip], y_grid_ext[::skip,::skip],
                qs_60_t[::skip,::skip], qn_60_t[::skip,::skip],
                color='k', scale=vmax*15, width=0.006)
ax[0, 2].set_title('$60^\\circ$')
ax[0, 2].set_xlim((0, 98))
ax[0, 2].set_xlabel('')
ax[0, 2].set_ylabel('')
ax[0, 2].tick_params(axis='x', which='both', labelbottom=False)
ax[0, 2].tick_params(axis='y', which='both', labelleft=False)
ax[0, 2].set_aspect('equal', adjustable='box')

cb = fig.colorbar(p2, cax=colorbar_ax)
cb.set_label('Transport rate magnitude [kg/m/s]')

x_section = x_grid_ext[cross_section_idx, :]
section_data = (
    (mag_0, qs_0_t),
    (mag_30, qs_30_t),
    (mag_60, qs_60_t),
)
section_ymin = min(
    np.min(values[cross_section_idx, :])
    for data_pair in section_data
    for values in data_pair
)
section_ymax = max(
    np.max(values[cross_section_idx, :])
    for data_pair in section_data
    for values in data_pair
)

for section_ax, angle, (magnitude, onshore) in zip(
    ax[1, :],
    ('0$^\\circ$', '30$^\\circ$', '60$^\\circ$'),
    section_data,
):
    section_ax.plot(x_section, magnitude[cross_section_idx, :], label='magnitude')
    section_ax.plot(x_section, onshore[cross_section_idx, :], label='onshore component')
    section_ax.axhline(0, color='k', linewidth=0.7)
    # section_ax.set_title(f'{angle}, alongshore index {cross_section_idx}')
    section_ax.set_xlim((0, 98))
    section_ax.set_ylim((0,0.01))
    section_ax.set_xlabel('cross-shore distance [m]')
    section_ax.grid(alpha=0.25)

ax[1, 0].set_ylabel('Transport rate [kg/m/s]')
ax[1, 1].set_ylabel('')
ax[1, 1].tick_params(axis='y', which='both', labelleft=False)
ax[1, 2].set_ylabel('')
ax[1, 2].tick_params(axis='y', which='both', labelleft=False)
# fig.tight_layout()

# fig.suptitle(f'Sediment transport rate at t={t[t_idx]:.0f} s')
```
