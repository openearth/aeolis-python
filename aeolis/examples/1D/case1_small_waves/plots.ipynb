```python
# read aeolis.nc file and plot the results
# using matplotlib and netCDF4 libraries
import matplotlib.pyplot as plt
from netCDF4 import Dataset
import numpy as np

# Open the netCDF file
nc_file = 'aeolis.nc'
data = Dataset(nc_file, 'r')
# Extract variables
x = np.squeeze(data.variables['x'][:])
z_end = np.squeeze(data.variables['zb'][-1,:])
z_0 = np.squeeze(data.variables['zb'][0,:])
data.close()

nc_file = 'aeolis_sts.nc'
data = Dataset(nc_file, 'r')
z_end_sts = np.squeeze(data.variables['zb'][-1,:])
# read rhoveg
rhoveg = np.squeeze(data.variables['rhoveg'][-1,:])
ustar = np.squeeze(data.variables['ustar'][-1,:])
Ct = np.squeeze(data.variables['Ct'][-1,:])



data.close()

# Plotting the results
# make a subplot with 7x3 size
plt.figure(figsize=(7, 3))
# plt.plot(x, z_end, label='Final Surface Elevation', color='blue')
plt.plot(x, z_0, label='Initial Surface Elevation', color='orange')
plt.plot(x, z_end_sts, label='Final Surface Elevation Sts', color='red')
plt.plot(x, z_end, label='Final Surface Elevation', color='blue')
plt.plot(x, rhoveg, label='Vegetation Density', color='green')
plt.title('Surface Elevation vs. Position')
plt.xlabel('Position (m)')
plt.ylabel('Surface Elevation (m)')
plt.grid()
plt.legend()
plt.show()


```

```python
# initate a new figure with a subpplot with 2 rows and 1 column
fig, axs = plt.subplots(3, 1, figsize=(7, 8))
# first subplot
axs[0].plot(x, z_0, label='Initial Surface Elevation', color='orange')
axs[0].plot(x, z_end_sts, label='Final Surface Elevation Sts', color='red')
axs[0].plot(x, z_end, label='Final Surface Elevation', color='blue')
axs[0].plot(x, rhoveg, label='Vegetation Density', color='green')
axs[0].set_title('Surface Elevation vs. Position')
axs[0].set_xlabel('Position (m)')
axs[0].set_ylabel('Surface Elevation (m)')
axs[0].grid()
axs[0].legend()
# second subplot
axs[1].plot(x,Ct, label='Transport Concentration \n(Ct [-])', color='gray')
axs[1].set_title('Vegetation Density vs. Position')
axs[1].set_xlabel('Position (m)')
axs[1].set_ylabel('Ct (-)')
axs[1].grid()
axs[1].legend()
# third subplot
axs[2].plot(x, ustar, label='Friction Velocity \n(ustar [m/s])', color='blue')
axs[2].set_title('Friction Velocity vs. Position')
axs[2].set_xlabel('Position (m)')
axs[2].set_ylabel('ustar (m/s)')
axs[2].grid()
axs[2].legend()
plt.tight_layout()
plt.show()
```
